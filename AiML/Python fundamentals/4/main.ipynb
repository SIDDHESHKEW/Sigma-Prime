```python
# An Basic class

class Student:

    def __init__(self,name,gpa):
        self.name = name
        self.gpa = gpa

    def __str__(self):
        return f"Student {self.name} got {self.gpa} gpa"

obj1 = Student("sidd", 10)

print(obj1)




```

## **Methods**
- Instance method
- Class method
- static method


```python

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram

    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)
l1.get_info()
```

```python
# Class method can only access class attributesabs

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram


    @classmethod                         # decorator
    def get_storage(cls):                # Class method
        print(f"Storage =",cls.storage_type)


    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)
l1.get_info()

l1.get_storage()
```

```python
# Static methods are stand alone functions used under a cclassmethod

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram

    @classmethod                         # decorator
    def get_storage(cls):                # Class method
        print(f"Storage =",cls.storage_type)


    @staticmethod
    def discount_cal(price,discount):
        print(f"Final price = {price - (price * discount/100)}")

    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)

l1.discount_cal(400000,10)
```
