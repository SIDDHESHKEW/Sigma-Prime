## **Methods**
- Instance method
- Class method
- static method


```python

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram

    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)
l1.get_info()
```

```python
# Class method can only access class attributesabs

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram


    @classmethod                         # decorator
    def get_storage(cls):                # Class method
        print(f"Storage =",cls.storage_type)


    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)
l1.get_info()

l1.get_storage()
```

```python
# Static methods are stand alone functions used under a cclassmethod

class Laptop:

    storage_type = "SSD"

    def __init__(self,storage, Ram):
        self.storage = storage
        self.Ram = Ram

    @classmethod                         # decorator
    def get_storage(cls):                # Class method
        print(f"Storage =",cls.storage_type)


    @staticmethod
    def discount_cal(price,discount):
        print(f"Final price = {price - (price * discount/100)}")

    def get_info(self):   # instance method
        print(f"Laptop with {self.Ram} GB Ram are okay and they must have {self.storage} GB's of storage.")

l1 = Laptop(225, 16)

l1.discount_cal(400000,10)
```

```python
# Practice problem

class store:

    count = 0

    def __init__(self,name,price):
        self.name = name
        self.price = price
        store.count += 1

    def items_count(cls):
        print(cls.count)

    def show_products(self):
        print(self.name)

    def discount(self,discount):
        print(f"Final price = {self.price-self.price * discount/100}")

s1 = store("Laptop", 50_000)
s2 = store("mobile", 5_000)

s1.show_products()
s1.discount(10)
s1.items_count()

```

```python
# Encapsulation - data hiding - the process of binding data and functions into a single unit while hiding internal state

class Bank:
    def __init__(self, name, balance, password):
        self.name = name  # public attribute
        self._balance = balance  # protected attribute  
        self.__password = password # private attribute


b1 = Bank("siddhesh", 50_000, 1972)

print(b1.name)
print(b1._balance)   # used _ to access it
# print(b1.__password) # gives error
# print(b1.password)  # gives error as private



```

```python
# how to access private attribute

# by internal getter method

class Bank:
    def __init__(self, name, balance, password):
        self.name = name  # public attribute
        self._balance = balance  # protected attribute  
        self.__password = password # private attribute

    def get_pass(self):
        return self.__password

b2 = Bank("siddhesh", 50_000, 1972)

print(b2.get_pass())

# or

print(b2._Bank__password)
```

```python
# Inheritence 

# - Single 
# - multilevel
# - multiple
# ----------------------------------------------------------------------


# Multilevel

class Employee:

    def __init__(self, salary):
        self.salary = salary

class Admin(Employee):

    def __init__(self, role):
        self.role = role

class Accountant(Admin):

    def __init__(self,role,salary):
        super().__init__(salary)
        Employee.__init__(self,role)

A1 = Accountant(50_000, "CA")

print(A1.salary, A1.role)

```

```python
# Multiple

class Student:

    subject = "Maths"

class teacher:

    def __init__(self,salary):
        self.salary = salary

class TA(Student,teacher):

    def __init__(self,salary):
        super().__init__(salary)

TA1 = TA(25_000)

print(TA1.salary,TA1.subject)

```

## **Diffrence Between Data Hiding and Abstration**

### Abstraction is the concept of hiding implementation details to expose only the essential features, focusing on what an object does.
- example - using chatgpt without knowing internal implementations
### Data hiding restricts direct access to an object's internal data, protecting it from unauthorized access and focusing on how data remains secure.
- example - hiding internal data by making it protected or private

```python
# Abstraction - Hiding complex inner workings and showing only the essential features.

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Lion(Animal):

    def sound(self):
        print("Roar....!!")

class Dog(Animal):

    def sound(self):
        print("Bark....!!")


l1 = Lion()
l2 = Dog()
l2.sound()
l1.sound()
```
