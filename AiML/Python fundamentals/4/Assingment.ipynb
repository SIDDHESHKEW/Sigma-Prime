```python
#1 

class BankAccount:

    def __init__(self,account_no,owner_name,balance):
        self.account_no = account_no
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self,amount):
        print(f"Previous balance = {self.balance}")
        self.balance += amount
        print(f"Current balance = {self.balance}")

    def withdraw(self,amount):
        print(f"Previous balance = {self.balance}")
        self.balance -= amount
        print(f"current balance = {self.balance}")

    def check_balance(self):
        print(f"Total balance = {self.balance}")


acc1 = BankAccount(123456789,"Siddhesh",895498560)

acc1.check_balance()
acc1.deposit(100000)
acc1.withdraw(600000000)
```

```python
#2 

class Book:

    def __init__(self, title, author):

        self.title = title
        self.author = author
        self.review_list = []

    def new_review(self,review:str):
        self.review_list.append(review)
        print("Review Added successfully")

    def count_review(self):
        print(len(self.review_list))

    def display_review(self):
        print(self.review_list)

b1 = Book("Harry Poter and hakwards","JK Roulings" )

b1.new_review("My sister is mad at this series")
b1.count_review()
b1.display_review()

```
