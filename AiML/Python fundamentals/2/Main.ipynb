```python
#2 

def even(a,b):
    list = []
    for i in range(a,b+1):
        if(i%2==0):
            list.append(i)
    
    print(f"All even numbers between (${a},${b}) =",list)

even(0,10)
```

```python
#9 

def prime_cheaker(n):
    prime = 0
    for i in (2,n-1):
        
        if(n%i == 0):
            print(f"{n} is not a prime number")
            prime+=1
            break
    if(prime==0):
        print(f"{n} is a prime number")
  

prime_cheaker(1033434394985573922)
```
