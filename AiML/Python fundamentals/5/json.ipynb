```python
import json
import os
```

## Writing to json


```python
with open("data.json","w") as f:
    py_obj = json.dump({"id": 1, "item": "apple", "price": 1.20},f )
```

```python
os.remove("data.json")
```

```python
with open("data.json","w") as f:
    py_obj = json.dump({"id": 1, "item": "apple", "price": 1.20},f, indent = 4 )
```

## Reading from json


```python
with open("data.json","r") as f:
    py_obj = json.load(f)
    print(py_obj)
```

## converting a string dict to js object 

```python
# js to py

string = '{"id": 1, "item": "apple", "price": 1.20, "present":true}'

py_dict = json.loads(string)
print(py_dict)
```

```python
# py to js

string = {"id": 1, "item": "apple", "price": 1.20, "present":True}
js_obj = json.dumps(string)
print(js_obj)
```
