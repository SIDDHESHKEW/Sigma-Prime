# File Operations

### Modes

| Mode | Description |
| :---: | :--- |
| **r** | #reading [default] |
| **w** | #writing, truncates file first |
| **x** | #creates new & open for writing |
| **a** | #writing, appends at end |
| **b** | #binary mode |
| **t** | #text mode [default] |
| **+** | #opens disk file for update(r & w) |
| **r+** | #opens file for both reading and writing (file must exist) |
| **w+** | #opens file for both reading and writing (truncates or creates file) |
| **a+** | #opens file for both reading and writing (appends to end or creates file) |


```python
with open("sample.txt","w+") as f:
    f.write("This text is written by main.ipynb")
    f.seek(0)                     # resets the pointer to index 0
    data = f.read()
    print(data)
```
