```python
#8 

l1 = [23,53,6,983,1]
l2 = [2,432,2,5,6]

intersect = set(l1).intersection(set(l2))

if(intersect == set()):
    print("No intersection between 2 lists")
else:
    print("Intersecting Element:",intersect)

```
